```python
def top_k(items, k):
    sorteditems=sorted(items,key=lambda x: x[1], reverse=True)
    return sorteditems[:k]

items=[("A",90),("B",95),("C",95),("D",85)]
k=2
print(top_k(items,k))

```

Q2

```python
def count_by_category(pairs):
    d={}
    for cat,word in pairs:
        if cat not in d:
            d[cat]={}
        if word not in d[cat]:
            d[cat][word]=0
        d[cat][word]+=1
        return d
    
pairs=[("sports","win"),("sports","win"),("tech","ai")]
print(count_by_category(pairs))
```

Q3

```python
def missing_number(nums):
    n = len(nums)
    total = n * (n + 1) // 2
    actual=sum(nums)
    return int(total -actual)

nums=[3,0,1]
print(missing_number(nums))
```

```python
def add_unique(x,seen=None):
    if seen is seen:
        seen=[]
    if x not in seen:
        seen.append(x)
    return seen
print(add_unique(1))
print(add_unique(2))
```
