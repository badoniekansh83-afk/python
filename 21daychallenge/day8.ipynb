### Q1 WAP to Generate all subsets of a list 

```python
def subset(arr, index=0, current=[]):
    if index == len(arr):
        print(current)
        return
    
    subset(arr, index + 1, current + [arr[index]])

    subset(arr, index +1, current)


arr=[1,2,3]
subset(arr)
```

### Q2 WAP to Check if two strings are anagrams using dictionary

```python
def is_anagram(s1,s2):
    if len(s1) != len(s2):
        return False
    
    frequency={}
    for i in s1:
        frequency[i]= frequency.get(i,0) + 1

    for i in s2:
        if i not in frequency or frequency[i] ==0:
            return False
        frequency[i] -=1

    return True

print(is_anagram("listen","silent"))
print(is_anagram("hello","coder"))
```
