Q1. You are given three rods named A (Source), B (Auxiliary), and C (Destination), along with n disks of different sizes placed on rod A. The disks are arranged in such a way that the largest disk is at the bottom and the smallest disk is on top.

Your task is to move all the disks from rod A to rod C using rod B as an auxiliary, following these rules:
1. Only one disk can be moved at a time.
2. A disk can only be placed on top of a larger disk or an empty rod.
3. You can only move the top disk of any rod.

```python
def tower_of_hanoi(n, source, auxiliary, destination):
    if n==1:
        print(f"Move disk 1 from {source} to {destination}")
        return
    tower_of_hanoi(n-1, source, destination, auxiliary)

    print(f"Move disk {n} from {source} to {destination}")

    tower_of_hanoi(n-1, auxiliary, source, destination)

n=3
tower_of_hanoi(n, 'A', 'B', 'C')
```

Q2. You are given a string consisting of distinct characters. Your task is to generate and print all possible permutations of the given string.A permutation is an arrangement of all the characters of the string in different possible orders.
Rules:
1. Each permutation must use all characters exactly once.
2. The order of characters matters (e.g., "ABC" and "BAC" are different).
3. No duplicate permutations should be printed (if characters are unique).

```python
def permutations(s, current=""):
    if len(s)==0:
        print(current)
        return
    
    for i in range(len(s)):
        char = s[i]
        remaining= s[:i] + s[i+1:]
        permutations(remaining, current + char)

permutations("ABC")
```
