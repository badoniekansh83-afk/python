#### Q1.  Create an abstract class Vehicle with abstract method start(). Implement it in Car.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key")

c=Car()
c.start()
```

#### Q2. Create a class with __del__() method and display message when object is destroyed.

```python
class Demo:
    def __init__(self,name):
        self.name = name
        print("Object created for", self.name)

    def __del__(self):
        print("Object destroyed for", self.name)

d= Demo("Ekansh")

del d
```
