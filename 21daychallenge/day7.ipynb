### Q1 WAP to find the First Repeating Element of the given list 

```python
def first_repeating(arr):
    freq={}

    for i in arr:
        freq[i]= freq.get(i,0)+1

    for i in arr:
        if freq[i]>1:
            return i
        
    return -1

arr=[10,5,3,4,5,6]
print(first_repeating(arr))
```

### Q2 WAP to Find Intersection of Two Arrays (Using Dictionary)

```python
def intersection(arr1, arr2):
    freq = {}
    result = []

    for i in arr1:
        freq[i] = freq.get(i, 0) + 1

    for i in arr2:
        if i in freq and freq[i] > 0:
            result.append(i)
            freq[i] -= 1  

    return result

arr1 = [1, 2, 2, 3, 4]
arr2 = [2, 2, 3, 5]

print(intersection(arr1, arr2))   
```
