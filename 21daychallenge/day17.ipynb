#### Q. Create a class Cart to manage shopping items with price and quantity. Include methods to add items, remove items, and calculate total price based on quantity. Display all items in the cart with their details.

```python
class Cart:
    def __init__(self):
        self.items = {}   
    def add_item(self, name, price, quantity):
        if name in self.items:
            self.items[name][1] += quantity
        else:
            self.items[name] = [price, quantity]
        print(name, "added to cart")

    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
            print(name, "removed from cart")
        else:
            print("Item not found")

    def total_price(self):
        total = 0
        for item in self.items:
            price, qty = self.items[item]
            total += price * qty
        print("Total Price:", total)

    def display_cart(self):
        print("\nCart Items:")
        for item in self.items:
            price, qty = self.items[item]
            print(item, "- Price:", price, "Quantity:", qty)


cart = Cart()

cart.add_item("Shoes", 2000, 2)
cart.add_item("Shirt", 1000, 1)
cart.add_item("Shoes", 2000, 1) 

cart.display_cart()
cart.total_price()
```
