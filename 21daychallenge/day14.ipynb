#### Q1 Create a class Employee with a constructor to initialize name and salary. Display the details using a method.

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    
    def display(self):
        print("Employee Name: ", self.name)
        print("Salary: ", self.salary)

a= Employee("Rahul",50000)

a.display()
```

#### Q2 Create a class BankAccount with: Public variable: name Private variable: __balance Create methods to deposit and display balance.

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Amount Deposited:",amount)

    def display_balance(self):
        print("Account Holder:", self.name)
        print("Balance:", self.__balance)


account1 = BankAccount("Ekansh",10000)

account1.deposit(500)

account1.display_balance()
```
