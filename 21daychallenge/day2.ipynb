Q1 Build a tool that: recognizes phone numbers in different formats (with/without country code, with dashes, dots, spaces), standardizes them to common format, validates Indian mobile numbers, and extracts area codes. Handle international formats.

```python
def clean_number(num):
    return num.replace(" ", "").replace("-", "").replace(".", "")


def normalize_number(num):
    num = clean_number(num)
    
    if num.startswith("+"):
        return num
    elif len(num) == 10:
        return "+91" + num
    elif num.startswith("91") and len(num) == 12:
        return "+" + num
    else:
        return num


def is_valid_indian(num):
    num = clean_number(num)
    
    if len(num) >= 10:
        last10 = num[-10:]
        return last10[0] in "6789" and len(last10) == 10
    return False


def extract_details(num):
    num = normalize_number(num)
    
    if num.startswith("+"):
        code = num[1:3]   
        local = num[-10:]
    else:
        code = "Unknown"
        local = num
    
    return code, local


if __name__ == "__main__":
    numbers = [
        "98765 43210",
        "91-9876543210",
        "+91.9876543210",
        "+1 1234567890"
    ]
    
    for n in numbers:
        print("Original:", n)
        print("Cleaned:", clean_number(n))
        print("Normalized:", normalize_number(n))
        print("Valid Indian:", is_valid_indian(n))
        print("Details (code, number):", extract_details(n))
        print("------")
```
