## Day 1  


Create a 'string_ops.py' module with functions to: count vowels/consonants, reverse words in sentence, check palindrome (ignoring case/spaces), count word frequency in text, and convert between camel_case, snake_case, and PascalCase. Test with a variety of inputs.

```python
def count_vowels_consonants(s):
    vowels = "aeiouAEIOU"
    v = 0
    c = 0
    
    for ch in s:
        if ch.isalpha():
            if ch in vowels:
                v += 1
            else:
                c += 1
    return v, c

def reverse_words(sentence):
    words = sentence.split()
    words.reverse()
    return " ".join(words)


def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]


def word_frequency(text):
    words = text.lower().split()
    freq = {}
    
    for word in words:
        if word in freq:
            freq[word] += 1
        else:
            freq[word] = 1
    return freq



def snake_to_camel(s):
    words = s.split('_')
    result = words[0]
    
    for w in words[1:]:
        result += w.capitalize()
    return result


def camel_to_snake(s):
    result = ""
    
    for ch in s:
        if ch.isupper():
            result += "_" + ch.lower()
        else:
            result += ch
    return result

def snake_to_pascal(s):
    words = s.split('_')
    return "".join(w.capitalize() for w in words)

if __name__ == "__main__":
    print(count_vowels_consonants("Hello World"))
    print(reverse_words("Python is easy"))
    print(is_palindrome("madam"))
    print(word_frequency("hello hello world"))
    
    print(snake_to_camel("my_name_is_python"))
    print(camel_to_snake("myNameIsPython"))
    print(snake_to_pascal("my_name_is_python"))
```
