## Sorting Algorithms

## 1. Bubble Sort

```python
def bubble_sort(arr):
    n=len(arr)

    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]= arr[j+1],arr[j]

    return arr

print(bubble_sort([2,43,62,1,68,34,24]))
```

## 2. Selection Sort

```python
def selection_sort(arr):
    n=len(arr)

    for i in range(n):
        min_index=i

    for j in range(i+1,n):
        if arr[j]< arr[min_index]:
            min_index+=1

    arr[i],arr[min_index]=arr[min_index],arr[i]

    return arr

print(selection_sort([4,23,1,65,35,87,45]))
```
