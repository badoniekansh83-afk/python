```python
class Employee:
    company= "ABC Ltd"
    def __init__(self,name,salary):
        self.name= name
        self.salary= salary
    
    def show_details(self):
        print(self.name,self.salary,self.company)

    @classmethod
    def change_company(cls, new_name):
        cls.company = new_name

emp1 = Employee("Ekansh",30000)
emp2 = Employee("Arun",40000)

emp1.show_details()
emp2.show_details()

Employee.change_company("XYZ ltd")

emp1.show_details()
emp2.show_details()


    
    
```

```python
class MathHelper:
    @staticmethod
    def is_even(num):
        if num%2 == 0:
            return True
        else:
            return False
        
print(MathHelper.is_even(8))
print(MathHelper.is_even(11))
```

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self._balance = balance

    def deposit(self,amount):
        self._balance +=amount
        print("Deposited:",amount)

    def withdraw(self,amount):
        if amount <=self._balance:
            self._balance -= amount
            print("Withdraw:",amount)
        else:
            print("Balance:",self._balance)
    
    def show_balance(self):
        print("Balance:",self._balance)

acc= BankAccount("Ekansh",1000)
acc.deposit(500)
acc.withdraw(300)
acc.show_balance()
```
